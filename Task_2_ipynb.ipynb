```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset
df = pd.read_csv("cleaned_data.csv")

# Display first 5 rows
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Statistical Summary

print("Statistical Summary:")
display(df.describe())

# Median values
print("\nMedian Values:")
print(df.median(numeric_only=True))

# Unique values in each column
print("\nNumber of Unique Values:")
print(df.nunique())
```

```python
# Histograms for numerical columns

plt.figure(figsize=(15, 4))

# Age distribution
plt.subplot(1, 3, 1)
plt.hist(df["age"], bins=10, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

# CGPA distribution
plt.subplot(1, 3, 2)
plt.hist(df["cgpa"], bins=10, edgecolor="black")
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")

# Placement distribution
plt.subplot(1, 3, 3)
plt.hist(df["placed"], bins=2, edgecolor="black")
plt.title("Placement Distribution")
plt.xlabel("Placed (0 = No, 1 = Yes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Bar charts for categorical columns

plt.figure(figsize=(15, 4))

# Gender distribution
plt.subplot(1, 3, 1)
df["gender"].value_counts().plot(kind="bar", edgecolor="black")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.xticks(rotation=0)

# Profession distribution
plt.subplot(1, 3, 2)
df["profession"].value_counts().plot(kind="bar", edgecolor="black")
plt.title("Profession Distribution")
plt.xlabel("Profession")
plt.ylabel("Count")
plt.xticks(rotation=0)

# City distribution
plt.subplot(1, 3, 3)
df["city"].value_counts().plot(kind="bar", edgecolor="black")
plt.title("City Distribution")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Scatter plots to identify relationships

plt.figure(figsize=(12, 5))

# Age vs CGPA
plt.subplot(1, 2, 1)
plt.scatter(df["age"], df["cgpa"], alpha=0.6)
plt.title("Age vs CGPA")
plt.xlabel("Age")
plt.ylabel("CGPA")

# Age vs Placement
plt.subplot(1, 2, 2)
plt.scatter(df["age"], df["placed"], alpha=0.6)
plt.title("Age vs Placement")
plt.xlabel("Age")
plt.ylabel("Placed (0 = No, 1 = Yes)")

plt.tight_layout()
plt.show()
```

```python
# Correlation analysis

correlation = df[["age", "cgpa", "placed"]].corr()

print("Correlation Matrix:")
display(correlation)

# Visualize correlation using heatmap

plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
