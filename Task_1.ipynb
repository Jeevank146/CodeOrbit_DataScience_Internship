```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("my_data (1).csv")

# Display the first 5 rows
df.head()
```

```python
# Check the number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Duplicate rows removed successfully.")
print("New number of rows:", df.shape[0])
```

```python
# Check missing values before cleaning

print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
# Handle missing values

# Fill missing values in text columns with the mode
text_columns = ["name", "city", "gender", "profession"]

for column in text_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# Fill missing values in numeric columns with the median
numeric_columns = ["age", "cgpa"]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing values handled successfully.")
```

```python
# Verify missing values after cleaning

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Calculate basic statistics for numeric columns

print("Mean values:")
print(df[["age", "cgpa"]].mean())

print("\nMedian values:")
print(df[["age", "cgpa"]].median())
```

```python
# Display value counts for categorical columns

print("Gender value counts:")
print(df["gender"].value_counts())

print("\nProfession value counts:")
print(df["profession"].value_counts())

print("\nPlaced value counts:")
print(df["placed"].value_counts())
```

```python
# Final verification of the cleaned dataset

print("Final dataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nFirst 5 rows of cleaned dataset:")
display(df.head())
```

```python
# Remove duplicates created after filling missing values

before = df.shape[0]

df = df.drop_duplicates()

after = df.shape[0]

print("Duplicate rows removed:", before - after)
print("Final number of rows:", after)
```

```python
# Save the cleaned dataset as a new CSV file

df.to_csv("cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully as cleaned_data.csv")
```
